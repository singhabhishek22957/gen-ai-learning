```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool 
import requests
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()

results = search_tool.invoke("Top new in India Today")
results
```

```python
llm = ChatOllama(
    model="gemma3:4b",
    temperature=0.7
)

# llm.invoke("Hi")
```

```python
from langchain_classic.agents import create_react_agent , AgentExecutor
from langsmith import Client
# from langchainhub import Client
hub = Client()
# from langchain_classic import hub
```

```python
# step 2 : pull the react( reasoning + action ) prompt from langchain hub

prompt = hub.pull_prompt("hwchase17/react", dangerously_pull_public_prompt=True)  # pulls the standard ReAct agent prompt

print(type(prompt))
```

```python
# step 3 create the ReAct agent manually with the pulled prompt 
agent = create_react_agent(
    llm=llm,
    tools=[search_tool],
    prompt=prompt
)
```

```python
# step 4: wrap it with AgentExecutor
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True
)
```

```python
response = agent_executor.invoke({"input":"give me info about Sri Narayan singh collage , Motihari "})
```
