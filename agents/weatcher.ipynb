```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
import requests
from langchain_community.tools import DuckDuckGoSearchRun
search_tool = DuckDuckGoSearchRun()
```

```python
@tool
def get_weather_data(city:str)->str:
    """this function the current weather data for a given city"""
    url = f"https://api.weatherstack.com/curent?access_key={api_key}&query={city}"
    res = requests.get(url)
    return res.json()
```

```python
from langchain_classic.agents import  create_react_agent , AgentExecutor
from langsmith import Client

hub = Client()
```

```python
prompt = hub.pull_prompt("hwchase17/react", dangerously_pull_public_prompt=True)  # pulls the standard ReAct agent prompt
```

```python
!ollama list
```

```python
llm = ChatOllama(
    model='gemma3:4b',
    temperature=0.7
)
```

```python
agent = create_react_agent(
    llm=llm,
    tools=[search_tool, get_weather_data],
    prompt=prompt
)
```

```python
agent_executor = AgentExecutor(
    agent= agent,
    tools = [search_tool, get_weather_data],
    verbose=True
)
```

```python
res = agent_executor.invoke({"input":" i want to visit in tamil naudu state so , give me 5 day plan where should i want to go  and i wan to cover at least 2 historic city  "})
print(res)
```
