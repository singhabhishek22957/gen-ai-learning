```python
from langchain_core.tools import tool
```

```python
# using @tool decorator
# less strict 
@tool
def multiply(a:int, b:int)-> int:
    """Multiply two numbers"""
    return a*b
```

```python
res = multiply.invoke({
    "a":3,
    "b":5
})
```

```python
print(res)
```
